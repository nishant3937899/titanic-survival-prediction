```python
sns.histplot(df['Age'],kde=True)
plt.title('Age distribution')
```

#The age is normally distributed


```python
df['Age'].fillna(df['Age'].mean() ,inplace=True)
```

```python
df['Embarked'].fillna(df['Embarked'].mode()[0],inplace=True)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.head()
```

```python
sns.countplot(data=df,x='Sex',hue='Survived')
plt.title('survival based on sex')
```

```python
gender=['male','female']
for i in gender:
    people_survived=len(df[(df['Sex'] == i) & (df['Survived']==1)])
    total_people=len(df[df['Sex']==i])
    print( f"the survial rate for {i} passenger is: ",(people_survived/total_people)*100)
```

# survial based on gender 
- on the basis of gender females passenger are more likely to survive than male passenger
- survial rate for male passenger 18.89%
- survial rate for female passenger is 74.20%

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='Age', hue='Survived', kde=True, bins=30)
plt.title('Age distribution by Survival')
plt.xlabel('Age')
plt.ylabel('Count')
```

# survial insight
- we can see that it is more likely for a woman and children to survive than a man 


```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Pclass', hue='Survived',)
plt.title('survival on basis of class')
plt.xlabel('class')
plt.ylabel('no. of peoples')
```

```python
for i in range(1,4):
    people_survived=len(df[(df['Pclass'] == i) & (df['Survived']==1)])
    total_people=len(df[df['Pclass']==i])
    print( f"the survial rate for class {i} is: ",(people_survived/total_people)*100)
```

# insights
- people in the 1st class are more likely to survive as their survial rate is 62.96% which is the highest
- people in the 2nd class have a much worse survival rate than people in 1st class which is 47.28%
- people in 3rd class have very low survial rate about 24.23% which is the lowest
